```python
activite_reconstituee = RunningActivities.from_fit_file(
    "activite_reconstituee.fit", segment_width=race.segment_width
)
activite_reconstituee.plot.track_map().show()
activite_reconstituee.data.head(5)
```

```python
from TrailPredictor.dataset import Dataset

dataset_pred = Dataset(activite_reconstituee.data.assign(activity_id=0))
dataset = Dataset.load(
    min_distance=40, max_distance=100, segment_width=activite_reconstituee.segment_width
)
plotter = dataset.plot
```

```python
from sklearn.model_selection import GroupKFold

dataset.target = "pace"
dataset_pred.target = "pace"
N_SPLITS = min(5, dataset.data["activity_id"].nunique())
RANDOM_STATE = 42
finite_rows = np.isfinite(dataset.target.to_numpy())
dataset._data = dataset._data.loc[finite_rows].reset_index(drop=True)

dataset.features = [f for f in dataset.features if f != "hearth_rate"]
dataset_pred.features = dataset.features

y = dataset.target
X = dataset.predictors
X_pred = dataset_pred.predictors
groups = dataset.data["activity_id"]
```

```python
from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import LassoCV, LinearRegression
from sklearn.pipeline import Pipeline

from regpath import (
    LassoPath,
    plot_lasso_path,
)

# create pipeline => preprocess + lasso + linear regression (on the features selected by lasso)
preprocess = dataset.get_preprocess()
cv = list(GroupKFold(n_splits=5).split(X, y, groups))
pipeline = Pipeline(
    [
        ("preprocess", preprocess),
        (
            "selection",
            SelectFromModel(LassoCV(cv=cv, random_state=RANDOM_STATE)),
        ),
        ("regression", LinearRegression()),
    ]
)
pipeline.fit(X, y)
```

```python
res = LassoPath.from_pipeline(pipeline, X, y, lasso_step="selection")
fig, _ = plot_lasso_path(res)
```

```python
from sklearn.model_selection import GroupKFold

# models = make_models()
# pred = cross_val_predict(models["linear"], X, y, groups=groups)
pred = pipeline.predict(X)

plt.figure(figsize=(12, 6))
plt.plot(pred)
plt.plot(dataset.target, alpha=0.5)

dataset.prediction = pred
dataset_pred.prediction = pipeline.predict(X_pred)

results = pd.concat(
    [
        dataset_pred.true_total_time,
        dataset_pred.predict_total_time(),
        # dataset.error_total_time(),
    ],
    axis=1,
)

# absolute error in minutes
results["error_total_time"] = np.abs(
    results["predict_total_time"] - results["true_total_time"]
)
results
```
